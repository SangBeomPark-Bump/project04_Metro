### 시작해보기

```python
from langchain.document_loaders import TextLoader, CSVLoader

loader = CSVLoader("final.csv")
documents = loader.load() 
```

```python
from langchain.embeddings import OpenAIEmbeddings, SentenceTransformerEmbeddings

# embedding_function = OpenAIEmbeddings(api_key=MYAPI)

embedding_function = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
from langchain_core.vectorstores import InMemoryVectorStore

vector_store = InMemoryVectorStore.from_documents(documents, embedding_function)
```

```python
# query = "Bring me the confusion where the name is HongDea the time is 15."
query = "Bring me the Myengdong and crowd density where the time is 17."
input_k = 2
results = vector_store.similarity_search(query, k=input_k)  # Get top 2 similar documents

str_results = []
for i in range(input_k):
    str_results.append(', '.join(results[i].page_content.split('\n')[1:]))

real_result = '\n'.join(str_results)
```

```python
from langchain_huggingface import HuggingFacePipeline, ChatHuggingFace
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

# 모델 ID 설정
model_id = "meta-llama/Llama-3.2-1B"

# 모델과 토크나이저 로드
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

tokenizer.chat_template = "{% for message in messages %}{{ message['role'] }}: {{ message['content'] }}{% endfor %}"

# 파이프라인 생성
pipe = pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=50)

# HuggingFacePipeline 객체 생성
hf = HuggingFacePipeline(pipeline=pipe)

```

```python
chat_model = ChatHuggingFace(llm=hf)

user_query = "The key to life is"
messages = [{"role": "user", "content": user_query}]

formatted_messages = tokenizer.apply_chat_template(messages, tokenize=False)

response = chat_model.invoke(formatted_messages)

# Print out the response content
# print(response['choices'][0]['message']['content'])
```
