```python
q_list = []
a_list = []
i_list = []
for _ in range(500):
    columns = range(5, 24)
    df = pd.DataFrame(np.random.uniform(0, 1.5, size=(len(names), 19)), columns=columns, index=names)
    # 소수점 셋째 자리에서 반올림
    value_df = df.round(2)

    name_index = random.randint(0, 2)
    luanch_time_index = random.randint(0, len(luanch_time)-1)
    times_wow_index = random.randint(0, len(times_wow)-1)



    name = names[name_index]
    csv_string = f': "{name}"'
    for i in range(5, 24):
        csv_string += f', {i} : {value_df.loc[name, i]}'

    time = times_wow[times_wow_index]

    question = luanch_time[luanch_time_index].replace("<name>", name).replace("<time>", time)

    congest_list = list(value_df.loc[name, times_wow_value[times_wow_index]])
    mean_congest =  value_df.loc[name, times_wow_value[times_wow_index]].mean().round(2)


    answer = f"The requested {time} congestion in {name} is "
    temp_list =[]
    for congestion, at_time in zip(congest_list, times_wow_value[times_wow_index]):
        temp_list.append(f"{at_time}:00 - {congestion}")
    answer += ', '.join(temp_list)
    answer += ". "

    answer += f"mean congestion is {mean_congest}. "

    if mean_congest < 0.5:
        answer += "This time frame is smooth, so there is no issue with visiting."
    elif mean_congest < 0.8:
        answer += "This time frame usually has moderate congestion, so you may occasionally experience some inconvenience while moving."
    else:
        answer += "This time frame is crowded, so we recommend visiting during a different time."

    q_list.append(question)
    a_list.append(answer)
    i_list.append(value_df.loc[name])


```
